# HealthConnect AI - Production Monitoring

Set up and verify production monitoring.

```python
import sys
sys.path.append('..')

from app.services.analytics_service import AnalyticsService
import asyncio

async def test_monitoring():
    # Initialize analytics service
    analytics = AnalyticsService()
    
    # Track some events
    await analytics.track_event("conversation_started", {"channel": "web"})
    await analytics.track_event("message_sent", {"intent": "appointment_booking"})
    await analytics.track_event("appointment_booked", {"type": "general"})
    
    # Get metrics
    metrics = await analytics.get_metrics()
    
    print("=== Current Metrics ===")
    for key, value in metrics.items():
        print(f"  {key}: {value}")
    
    # Generate report
    report = await analytics.generate_report("daily")
    print(f"\n=== Report ===")
    print(f"  Generated at: {report['generated_at']}")
    print(f"  Period: {report['period']}")

asyncio.run(test_monitoring())

print("\n=== Monitoring Endpoints ===")
print("Prometheus: http://localhost:9090")
print("Grafana: http://localhost:3000")
print("API Metrics: http://localhost:8000/metrics")
print("Health Check: http://localhost:8000/health")
```
