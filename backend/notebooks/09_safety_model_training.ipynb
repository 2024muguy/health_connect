# HealthConnect AI - Safety Model Training

Train safety classification model.

```python
import sys
sys.path.append('..')

from training.data_preparation import DataPreparation
from training.safety_classifier import SafetyClassifierTrainer

# Prepare data
data_prep = DataPreparation()
training_data = data_prep.generate_safety_training_data()

print(f"Total training examples: {len(training_data)}")

# Split data
train_data, val_data, test_data = data_prep.split_data(
    training_data,
    train_ratio=0.8,
    val_ratio=0.1,
    test_ratio=0.1,
)

# Train model
trainer = SafetyClassifierTrainer()
results = trainer.train(train_data, val_data)

print(f"\nTraining Results:")
print(f"  Train Accuracy: {results['train_accuracy']:.4f}")
print(f"  Val Accuracy: {results.get('val_accuracy', 'N/A')}")

# Test predictions
test_queries = [
    "How do I book an appointment?",
    "I have a headache, what should I take?",
    "I'm having chest pain!",
]

print("\nTest Predictions:")
for query in test_queries:
    prediction = trainer.predict(query)
    print(f"  '{query}' -> {prediction['safety_category']} (risk: {prediction['risk_score']})")

# Save model
trainer.save()
print("\nModel saved!")
```
