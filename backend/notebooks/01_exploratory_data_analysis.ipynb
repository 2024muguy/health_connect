# HealthConnect AI - Exploratory Data Analysis

This notebook explores the HealthConnect Knowledge Base and Appointment Data to understand the data structure and identify patterns.

```python
# Import required libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

# Set plotting style
plt.style.use('seaborn-v0_8-darkgrid')
sns.set_palette('husl')

# Display settings
pd.set_option('display.max_columns', 50)
pd.set_option('display.max_rows', 100)
```

```python
# Load appointment data
data_path = Path('../data/raw/HealthConnect_Appointment_Data.csv')

if data_path.exists():
    df = pd.read_csv(data_path)
    print(f"Dataset shape: {df.shape}")
    print(f"\nColumns: {df.columns.tolist()}")
    print(f"\nData types:\n{df.dtypes}")
else:
    print("Appointment data file not found. Please download and place in data/raw/")
    df = None
```

```python
# Basic statistics
if df is not None:
    print("=== Basic Statistics ===")
    print(df.describe())
    
    print("\n=== Missing Values ===")
    print(df.isnull().sum())
    
    print("\n=== Target Variable (No-Show) ===")
    if 'no_show' in df.columns:
        print(df['no_show'].value_counts())
        print(f"\nNo-show rate: {df['no_show'].mean() * 100:.2f}%")
```

## Knowledge Base Analysis

Explore the Knowledge Base structure using python-docx.

```python
# Load Knowledge Base
from docx import Document

kb_path = Path('../data/raw/HealthConnect_Clinic_Knowledge_Base.docx')

if kb_path.exists():
    doc = Document(kb_path)
    
    # Extract all paragraphs
    paragraphs = [p.text for p in doc.paragraphs if p.text.strip()]
    
    print(f"Total paragraphs: {len(paragraphs)}")
    print(f"\nFirst 20 paragraphs:")
    for i, para in enumerate(paragraphs[:20], 1):
        print(f"{i}. {para[:100]}")
else:
    print("Knowledge Base file not found")
    paragraphs = []
```
