# HealthConnect AI - Model Fine-Tuning

Fine-tune embedding models for domain adaptation.

```python
import sys
sys.path.append('..')

from training.embedding_fine_tuning import EmbeddingFineTuner

# Initialize fine-tuner
tuner = EmbeddingFineTuner()

# Fine-tune model
results = tuner.fine_tune(
    base_model="all-MiniLM-L6-v2",
    epochs=3,
    batch_size=16,
)

print(f"Fine-tuning results: {results}")

# Evaluate
eval_results = tuner.evaluate()
print(f"Evaluation: {eval_results}")
```
