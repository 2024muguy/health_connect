# HealthConnect AI - Intent Classifier Training

Train intent classification model.

```python
import sys
sys.path.append('..')

from training.data_preparation import DataPreparation
from training.intent_classifier import IntentClassifierTrainer
from training.evaluator import ModelEvaluator

# Prepare data
data_prep = DataPreparation()
training_data = data_prep.generate_intent_training_data()

print(f"Total training examples: {len(training_data)}")

# Split data
train_data, val_data, test_data = data_prep.split_data(
    training_data,
    train_ratio=0.8,
    val_ratio=0.1,
    test_ratio=0.1,
)

print(f"Train: {len(train_data)}, Val: {len(val_data)}, Test: {len(test_data)}")

# Train model
trainer = IntentClassifierTrainer()
results = trainer.train(train_data, val_data)

print(f"\nTraining Results:")
print(f"  Train Accuracy: {results['train_accuracy']:.4f}")
print(f"  Val Accuracy: {results.get('val_accuracy', 'N/A')}")

# Save model
trainer.save()
print("\nModel saved!")
```
