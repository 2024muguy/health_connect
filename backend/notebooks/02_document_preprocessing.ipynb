# HealthConnect AI - Document Preprocessing

Clean and preprocess the Knowledge Base document for RAG pipeline.

```python
import re
import unicodedata
from pathlib import Path
from docx import Document

def clean_text(text):
    """Clean and normalize text"""
    # Unicode normalization
    text = unicodedata.normalize('NFKC', text)
    
    # Remove control characters
    text = ''.join(char for char in text if ord(char) >= 32 or char == '\n')
    
    # Normalize quotes
    text = text.replace('\u2018', "'").replace('\u2019', "'")
    text = text.replace('\u201c', '"').replace('\u201d', '"')
    
    # Remove extra whitespace
    text = re.sub(r'[ \t]+', ' ', text)
    text = re.sub(r'\n{3,}', '\n\n', text)
    
    return text.strip()

# Load and clean Knowledge Base
kb_path = Path('../data/raw/HealthConnect_Clinic_Knowledge_Base.docx')

if kb_path.exists():
    doc = Document(kb_path)
    raw_text = '\n'.join([p.text for p in doc.paragraphs if p.text.strip()])
    cleaned_text = clean_text(raw_text)
    
    # Save cleaned text
    output_path = Path('../data/processed/knowledge_base_cleaned.txt')
    output_path.parent.mkdir(parents=True, exist_ok=True)
    output_path.write_text(cleaned_text)
    
    print(f"Original length: {len(raw_text)} characters")
    print(f"Cleaned length: {len(cleaned_text)} characters")
    print(f"Saved to: {output_path}")
else:
    print("Knowledge Base file not found")
```
