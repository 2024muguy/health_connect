# HealthConnect AI - Chunking Strategies

Experiment with different chunking strategies for RAG pipeline.

```python
import sys
sys.path.append('..')

from app.rag.chunking import ChunkingEngine, ChunkingStrategy
from pathlib import Path

# Load cleaned text
text_path = Path('../data/processed/knowledge_base_cleaned.txt')

if text_path.exists():
    text = text_path.read_text()
    print(f"Loaded text: {len(text)} characters")
    
    # Initialize chunking engine
    engine = ChunkingEngine()
    
    # Test different strategies
    strategies = [
        ChunkingStrategy.FIXED_SIZE,
        ChunkingStrategy.SEMANTIC,
        ChunkingStrategy.SENTENCE_AWARE,
        ChunkingStrategy.HYBRID,
    ]
    
    for strategy in strategies:
        chunks = engine.chunk_document(
            text,
            document_id="test",
            chunk_size=512,
            overlap=77,
            strategy=strategy,
        )
        print(f"\n{strategy.value}: {len(chunks)} chunks")
        if chunks:
            print(f"  First chunk: {chunks[0].text[:100]}...")
            print(f"  Avg tokens: {sum(c.token_count for c in chunks) / len(chunks):.0f}")
else:
    print("Cleaned text not found. Run notebook 02 first.")
```
