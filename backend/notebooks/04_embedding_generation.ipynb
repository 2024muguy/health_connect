# HealthConnect AI - Embedding Generation

Generate embeddings for knowledge base chunks.

```python
import sys
sys.path.append('..')

from app.rag.embedding import EmbeddingGenerator
from pathlib import Path
import json

# Test embedding generation
generator = EmbeddingGenerator()

test_texts = [
    "HealthConnect Clinic is located at 123 Main Street.",
    "Our opening hours are Monday to Friday 8am-6pm.",
    "To book an appointment, please call our scheduling line.",
]

print(f"Embedding model: {generator.model_name}")
print(f"Embedding dimension: {generator.dimension}")

# Generate embeddings (requires API key)
import asyncio

async def generate():
    embeddings = await generator.embed_texts(test_texts)
    return embeddings

try:
    embeddings = asyncio.run(generate())
    print(f"\nGenerated {len(embeddings)} embeddings")
    print(f"Embedding shape: {len(embeddings[0])} dimensions")
    
    # Test similarity
    sim = generator.cosine_similarity(embeddings[0], embeddings[1])
    print(f"Similarity between text 1 and 2: {sim:.4f}")
except Exception as e:
    print(f"Embedding generation failed: {e}")
    print("Note: Requires API key in .env file")
```
