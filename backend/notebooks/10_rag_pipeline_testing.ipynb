# HealthConnect AI - RAG Pipeline Testing

End-to-end RAG pipeline testing.

```python
import sys
sys.path.append('..')

from app.rag.document_ingestion import DocumentIngestionPipeline
from app.rag.retriever import HybridRetriever
import asyncio

async def test_pipeline():
    # Ingest document
    pipeline = DocumentIngestionPipeline()
    result = await pipeline.ingest_knowledge_base(
        "../data/raw/HealthConnect_Clinic_Knowledge_Base.docx"
    )
    print(f"Ingestion result: {result}")
    
    # Test retrieval
    retriever = HybridRetriever()
    
    test_queries = [
        "Where is the clinic located?",
        "How do I book an appointment?",
        "What are your hours?",
    ]
    
    for query in test_queries:
        results = await retriever.retrieve(query, top_k=3)
        print(f"\nQuery: {query}")
        for i, result in enumerate(results, 1):
            print(f"  {i}. Score: {result.score:.3f} - {result.text[:80]}...")

asyncio.run(test_pipeline())
```
